```python
marks = int(input("Enter student marks: "))

if marks >= 50:
    print("Student is Pass")
    print("Final Marks:", marks)
else:
    grace_marks = 50 - marks
    marks = marks + grace_marks
    print("Student is Pass with Grace Marks")
    print("Final Marks after Grace:", marks)
```
